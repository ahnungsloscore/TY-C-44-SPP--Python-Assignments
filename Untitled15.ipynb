```python
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms as T

tf = T.Compose([T.Resize((224,224)), T.ToTensor(),
                T.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])])
class DS(Dataset):
    def __init__(s, paths): s.p = list(paths)
    def __len__(s): return len(s.p)
    def __getitem__(s, i): return tf(Image.open(s.p[i]).convert("RGB"))

loader = DataLoader(DS(meta["path"]), batch_size=64, num_workers=2)
preds, probs = [], []
with torch.no_grad():
    for x in loader:
        out = model(x.to("cuda")).softmax(1).cpu()
        preds += (out.argmax(1) == ECZEMA).int().tolist()
        probs += out[:, ECZEMA].tolist()
meta["pred"], meta["p_eczema"] = preds, probs
meta.to_csv("/content/dermacon_preds.csv", index=False)
print("done:", len(meta), "photos; predicted eczema:", sum(preds))
```

```python
import pandas as pd

def wilson(k, n, z=1.96):
    if n == 0: return (0, 0)
    p = k/n; d = 1 + z*z/n
    c = (p + z*z/(2*n))/d
    h = z*((p*(1-p)/n + z*z/(4*n*n))**0.5)/d
    return round(c-h, 2), round(c+h, 2)

def table(label_col):
    rows = []
    for g, d in [("ALL", meta)] + list(meta.groupby("tone")):
        tp=((d.pred==1)&(d[label_col]==1)).sum(); fn=((d.pred==0)&(d[label_col]==1)).sum()
        tn=((d.pred==0)&(d[label_col]==0)).sum(); fp=((d.pred==1)&(d[label_col]==0)).sum()
        rows.append({"tone": g, "eczema_photos": tp+fn,
                     "sensitivity": round(tp/max(tp+fn,1), 2),
                     "sens_range_95%": wilson(tp, tp+fn),
                     "specificity": round(tn/max(tn+fp,1), 2),
                     "precision": round(tp/max(tp+fp,1), 2),
                     "false_alarms": fp})
    return pd.DataFrame(rows).to_string(index=False)

print("STRICT eczema definition")
print(table("label"))

# broad definition: anything with eczema or dermatitis in the name
meta["label_broad"] = meta["Disease_label"].astype(str).str.contains("eczema|dermatitis", case=False).astype(int)
print("\nBROAD eczema/dermatitis definition")
print(table("label_broad"))
```

```python
from sklearn.metrics import roc_auc_score
for g, d in [("ALL", meta)] + list(meta.groupby("tone")):
    print(g, "| AUC strict:", round(roc_auc_score(d.label, d.p_eczema), 3),
          "| AUC broad:", round(roc_auc_score(d.label_broad, d.p_eczema), 3),
          "| % called eczema:", round(d.pred.mean()*100, 1))
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# If Colab restarted and forgot the results, uncomment the next line:
# meta = pd.read_csv("/content/dermacon_preds.csv")

order = ["FST 3", "FST 4", "FST 5-6"]
rows = []
for g in order:
    d = meta[meta.tone == g]
    acc  = (d.pred == d.label).mean() * 100
    sens = ((d.pred==1)&(d.label==1)).sum() / max((d.label==1).sum(), 1) * 100
    spec = ((d.pred==0)&(d.label==0)).sum() / max((d.label==0).sum(), 1) * 100
    rows.append({"Skin tone": g, "Photos": len(d), "Accuracy %": round(acc, 1),
                 "Eczema caught %": round(sens, 1), "Non-eczema cleared %": round(spec, 1)})
res = pd.DataFrame(rows)
print(res.to_string(index=False))
print("\nOverall accuracy: %.1f%%" % ((meta.pred == meta.label).mean() * 100))

x = np.arange(len(order)); w = 0.27
fig, ax = plt.subplots(figsize=(9, 5))
for i, (col, color) in enumerate([("Accuracy %", "#4C72B0"),
                                  ("Eczema caught %", "#DD8452"),
                                  ("Non-eczema cleared %", "#55A868")]):
    bars = ax.bar(x + (i-1)*w, res[col], w, label=col, color=color)
    ax.bar_label(bars, fmt="%.0f%%", fontsize=9)
ax.set_xticks(x); ax.set_xticklabels(order)
ax.set_ylim(0, 110); ax.set_ylabel("Percent")
ax.set_title("Eczema AI (clinical-skin-condition-v1) on DermaCon-IN, by skin tone")
ax.legend(loc="upper center", ncol=3, bbox_to_anchor=(0.5, -0.08))
plt.tight_layout(); plt.show()
```
